1. What is Python, and why is it popular

Python is a high-level, versatile programming language known for its simplicity and readability. It's popular due to its wide range of applications, from web development to data science and AI.

2. What is an interpreter in Python

An interpreter in Python executes code line by line, converting it into machine language at runtime. It helps in quick testing and debugging, making Python easy to use and learn.

3. What are pre-defined keywords in Python

Pre-defined keywords in Python are reserved words that have special meanings and purposes, like `if`, `while`, `def`, and `return`. They cannot be used as variable names or identifiers.

4.  Can keywords be used as variable names

No, keywords cannot be used as variable names in Python because they are reserved for specific syntax and functions. Using them as variable names will result in a syntax error.

5. What is mutability in Python

Mutability in Python refers to whether an object’s value can be changed after it is created. Mutable objects (like lists) can be altered, while immutable ones (like strings and tuples) cannot.

6. Why are lists mutable, but tuples are immutable

Lists are mutable because they are designed to store dynamic data that can change over time. Tuples are immutable to ensure data integrity and are often used for fixed collections of items.

7.  What is the difference between “==” and “is” operators in Python

The `==` operator checks if two variables have the same value. The `is` operator checks if they refer to the same object in memory.

8.  What are logical operators in Python

Logical operators in Python are used to combine conditional statements and include `and`, `or`, and `not`. They return Boolean values based on the logic between expressions.

9. What is type casting in Python

Type casting in Python is the process of converting one data type into another, like from `int` to `float`. It helps ensure correct operations between different data types.

10. What is the difference between implicit and explicit type casting

Implicit type casting is done automatically by Python when converting compatible data types. Explicit type casting is manually done by the programmer using functions like `int()`, `float()`, or `str()`.

11. What is the purpose of conditional statements in Python

Conditional statements in Python, like `if`, `elif`, and `else`, allow the program to execute different code blocks based on specific conditions. They enable decision-making and control the flow of the program.

12. How does the elif statement work

The `elif` statement in Python allows checking multiple conditions after an `if` statement. If the `if` condition is false, it checks the `elif` condition, executing its block if true.

13. What is the difference between for and while loops

A `for` loop iterates over a sequence (like a list or range) for a fixed number of times. A `while` loop continues to execute as long as a given condition remains true.

14.  Describe a scenario where a while loop is more suitable than a for loop.

A `while` loop is more suitable when the number of iterations is unknown and depends on a condition, such as reading data until the end of a file is reached. It's ideal when the loop should continue until a specific event or condition occurs.

4. Write a program that checks if a given word is a Python keyword.

```python
import keyword

# Function to check if a word is a Python keyword
def is_keyword(word):
    if keyword.iskeyword(word):
        print(f"'{word}' is a Python keyword.")
    else:
        print(f"'{word}' is not a Python keyword.")

# Input from the user
word = input("Enter a word to check if it's a Python keyword: ")

# Check if the word is a Python keyword
is_keyword(word)
```

6.  Write a function to demonstrate the behavior of mutable and immutable arguments

```python
# Function to demonstrate mutable and immutable arguments
def demo_mutable_immutable(arg1, arg2):
    # Modifying the mutable argument (list)
    arg1.append(100)

    # Trying to modify the immutable argument (integer)
    arg2 += 10

    print("Inside function:")
    print("Mutable argument (list) after modification:", arg1)
    print("Immutable argument (integer) after modification:", arg2)

# Calling the function with a mutable argument (list) and an immutable argument (integer)
my_list = [1, 2, 3]
my_int = 5

print("Before function call:")
print("Mutable argument (list):", my_list)
print("Immutable argument (integer):", my_int)

# Calling the function
demo_mutable_immutable(my_list, my_int)

# Checking the values after function call
print("\nAfter function call:")
print("Mutable argument (list):", my_list)  # List is modified
print("Immutable argument (integer):", my_int)  #_
```

7.  Write a program that performs basic arithmetic operations on two user-input numbers.



```python
# Function to perform basic arithmetic operations
def arithmetic_operations():
    # Taking user input for two numbers
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))

    # Performing the operations
    print(f"\nAddition: {num1} + {num2} = {num1 + num2}")
    print(f"Subtraction: {num1} - {num2} = {num1 - num2}")
    print(f"Multiplication: {num1} * {num2} = {num1 * num2}")

    # Handling division by zero
    if num2 != 0:
        print(f"Division: {num1} / {num2} = {num1 / num2}")
    else:
        print("Division: Cannot divide by zero!")

# Call the function to execute
arithmetic_operations()
```

9. Write a Python program to convert user input from string to integer, float, and boolean types.

```python
# Function to convert user input to different data types
def convert_input():
    # Taking user input as a string
    user_input = input("Enter a value: ")

    # Converting to integer
    try:
        int_value = int(user_input)
        print(f"Converted to Integer: {int_value}")
    except ValueError:
        print("Input cannot be converted to Integer.")

    # Converting to float
    try:
        float_value = float(user_input)
        print(f"Converted to Float: {float_value}")
    except ValueError:
        print("Input cannot be converted to Float.")

    # Converting to boolean
    # A non-empty string will be converted to True, an empty string will be False
    bool_value = bool(user_input)
    print(f"Converted to Boolean: {bool_value}")

# Call the function to execute
convert_input()
```

10.  Write code to demonstrate type casting with list elements.

```python
# Function to demonstrate type casting with list elements
def type_casting_with_list():
    # Original list with mixed types (string, float, and integer)
    mixed_list = ['10', '20.5', 30]

    # Converting all elements to integers
    int_list = [int(float(item)) if isinstance(item, str) else int(item) for item in mixed_list]
    print("List with elements converted to integers:", int_list)

    # Converting all elements to floats
    float_list = [float(item) for item in mixed_list]
    print("List with elements converted to floats:", float_list)

    # Converting all elements to strings
    str_list = [str(item) for item in mixed_list]
    print("List with elements converted to strings:", str_list)

# Call the function to execute
type_casting_with_list()
```

14. Write a program to reverse a string using a while loop.

```python
# Function to reverse a string using a while loop
def reverse_string():
    # Taking user input for the string
    user_string = input("Enter a string: ")

    # Initializing variables
    reversed_string = ""
    index = len(user_string) - 1  # Starting from the last character

    # While loop to reverse the string
    while index >= 0:
        reversed_string += user_string[index]
        index -= 1

    # Output the reversed string
    print("Reversed string:", reversed_string)

# Call the function to execute
reverse_string()
```

15. Write a Python program to calculate the factorial of a number provided by the user using a while loop

```python
# Function to calculate factorial using a while loop
def calculate_factorial():
    # Taking user input
    num = int(input("Enter a number: "))

    # Initialize factorial as 1
    factorial = 1

    # Check if the number is negative
    if num < 0:
        print("Factorial is not defined for negative numbers.")
    else:
        # While loop to calculate factorial
        while num > 1:
            factorial *= num
            num -= 1

        # Output the result
        print("The factorial is:", factorial)

# Call the function to execute
calculate_factorial()
```
